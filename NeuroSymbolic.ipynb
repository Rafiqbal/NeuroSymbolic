```python
# Function to load and preprocess images (resize and normalize)
def load_and_preprocess_image(image_path):
    try:
        # Check if the image exists
        if not os.path.exists(image_path):
            print(f"File not found: {image_path}")
            return None
        
        # Open and resize image using PIL
        img = Image.open(image_path).resize((224, 224))  # Resize to 224x224
        img_array = np.array(img)  # Convert image to numpy array
        img_array = img_array / 255.0  # Normalize pixel values to [0, 1]
        return img_array
    except Exception as e:
        print(f"Error loading image {image_path}: {e}")
        return None
```

```python
# Load and preprocess images for the training data
train_images = []
train_labels = []
```

```python
# Check if images are being read correctly
print("\nReading images from the directory...")
for filename in os.listdir(train_images_dir):
    if filename.endswith(('.jpeg', '.jpg', '.png')):  # Handle common image extensions
        print(f"Found image: {filename}")  # Debug: Print filenames being read

        # Extract the image ID (e.g., '10' from '10_left.jpeg')
        image_id = filename.split('_')[0]  
        
        if image_id in labels:
            # Add .jpeg extension to the image_id from the CSV
            expected_filename_left = f"{image_id}_left.jpeg"
            expected_filename_right = f"{image_id}_right.jpeg"
            
            # Check if the image matches either the left or right
            if filename == expected_filename_left or filename == expected_filename_right:
                image_path = os.path.join(train_images_dir, filename)
                print(f"Expected image path: {image_path}")  # Debug: Print expected image path
                
                img_array = load_and_preprocess_image(image_path)
                if img_array is not None:
                    train_images.append(img_array)
                    train_labels.append(labels[image_id])
                else:
                    print(f"Error in processing image: {image_path}")  # Debug: Processing error
                
```

```python
# Convert to numpy arrays
train_images = np.array(train_images)
train_labels = np.array(train_labels)
```
